```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

file_path = "../Data/breast-cancer.csv"
df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print(df.head())
```

```python
target_col = "diagnosis"

X = df.drop(columns=[target_col])
y = df[target_col]

print("\nNumber of features:", X.shape[1])

# In this assignment, every 0 is treated as a missing value.
X_missing = X.replace(0, np.nan)

print("\nMissing values after replacing 0 with NaN:")
print(X_missing.isna().sum())
```

```python
mean_imputer = SimpleImputer(strategy="mean")

X_mean = pd.DataFrame(
    mean_imputer.fit_transform(X_missing),
    columns=X.columns
)

median_imputer = SimpleImputer(strategy="median")

X_median = pd.DataFrame(
    median_imputer.fit_transform(X_missing),
    columns=X.columns
)
```

```python
print("Remaining missing values (Mean):",
      X_mean.isna().sum().sum())

print("Remaining missing values (Median):",
      X_median.isna().sum().sum())
```

```python
df_mean = X_mean.copy()
df_mean[target_col] = y.values

df_median = X_median.copy()
df_median[target_col] = y.values

print("\nMean-imputed dataset:")
print(df_mean.head())

print("\nMedian-imputed dataset:")
print(df_median.head())
```

```python
mean_output = "breast-cancer-mean-imput.csv"
median_output = "breast-cancer-median-imput.csv"

df_mean.to_csv(mean_output, index=False)
df_median.to_csv(median_output, index=False)

print(f"Mean-imputed dataset saved to: {mean_output}")
print(f"Median-imputed dataset saved to: {median_output}")
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

standard_scaler = StandardScaler()

X_mean_standard = pd.DataFrame(
    standard_scaler.fit_transform(X_mean),
    columns=X_mean.columns
)

X_median_standard = pd.DataFrame(
    standard_scaler.fit_transform(X_median),
    columns=X_median.columns
)

df_mean_standard = X_mean_standard.copy()
df_mean_standard[target_col] = y.values

df_median_standard = X_median_standard.copy()
df_median_standard[target_col] = y.values

df_mean_standard.to_csv(
    "breast-cancer-mean-standard.csv",
    index=False
)

df_median_standard.to_csv(
    "breast-cancer-median-standard.csv",
    index=False
)

print("Saved:")
print("  breast-cancer-mean-standard.csv")
print("  breast-cancer-median-standard.csv")
```

```python
minmax_scaler = MinMaxScaler()

X_mean_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(X_mean),
    columns=X_mean.columns
)

X_median_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(X_median),
    columns=X_median.columns
)

df_mean_minmax = X_mean_minmax.copy()
df_mean_minmax[target_col] = y.values

df_median_minmax = X_median_minmax.copy()
df_median_minmax[target_col] = y.values

df_mean_minmax.to_csv(
    "breast-cancer-mean-minmax.csv",
    index=False
)

df_median_minmax.to_csv(
    "breast-cancer-median-minmax.csv",
    index=False
)

print("Saved:")
print("  breast-cancer-mean-minmax.csv")
print("  breast-cancer-median-minmax.csv")
```
