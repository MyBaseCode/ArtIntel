```python
import pandas as pd

from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

TARGET = "diagnosis"

df = pd.read_csv("breast-cancer-mean-minmax.csv")

X = df.drop(columns=[TARGET])
y = df[TARGET]

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

models = {
    "KNN (k=3)": KNeighborsClassifier(n_neighbors=3),

    "KNN (k=9)": KNeighborsClassifier(n_neighbors=9),

    "Decision Tree (depth=2)": DecisionTreeClassifier(
        max_depth=2,
        random_state=42
    ),

    "Decision Tree (depth=8)": DecisionTreeClassifier(
        max_depth=8,
        random_state=42
    )
}

results = {}
```

```python
for name, model in models.items():
    scores = cross_val_score(
        model,
        X,
        y,
        cv=cv,
        scoring="accuracy"
    )
    
    results[name] = scores * 100

cv_results = pd.DataFrame(results)

cv_results.loc["Mean"] = cv_results.mean()
cv_results.loc["Std Dev"] = cv_results.std()

cv_results.index = [
    "Fold 1",
    "Fold 2",
    "Fold 3",
    "Fold 4",
    "Fold 5",
    "Fold 6",
    "Fold 7",
    "Fold 8",
    "Fold 9",
    "Fold 10",
    "Mean",
    "Std Dev"
]

print(cv_results.round(2))
```
