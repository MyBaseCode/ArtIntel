```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

TARGET = "diagnosis"
TEST_SIZE = 0.30
RANDOM_STATE = 42
K = 5
MAX_DEPTH = 8

datasets = {
    "No Normalization": "breast-cancer-mean-imput.csv",
    "Standardization": "breast-cancer-mean-standard.csv",
    "Min-Max Scaling": "breast-cancer-mean-minmax.csv"
}

results = []

for dataset_name, file in datasets.items():

    df = pd.read_csv(file)

    X = df.drop(columns=[TARGET])
    y = df[TARGET]

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE,
        stratify=y
    )

    # ---------------------
    # KNN
    # ---------------------

    knn = KNeighborsClassifier(
        n_neighbors=K
    )

    knn.fit(X_train, y_train)

    knn_pred = knn.predict(X_test)

    knn_acc = accuracy_score(y_test, knn_pred)

    # ---------------------
    # Decision Tree
    # ---------------------

    dt = DecisionTreeClassifier(
        max_depth=MAX_DEPTH,
        random_state=RANDOM_STATE
    )

    dt.fit(X_train, y_train)

    dt_pred = dt.predict(X_test)

    dt_acc = accuracy_score(y_test, dt_pred)

    results.append({
        "Dataset": dataset_name,
        "KNN Accuracy (%)": knn_acc * 100,
        "Decision Tree Accuracy (%)": dt_acc * 100
    })

results_df = pd.DataFrame(results)

print(results_df.to_string(index=False))
```
