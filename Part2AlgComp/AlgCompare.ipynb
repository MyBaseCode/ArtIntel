```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier, RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

TARGET = "diagnosis"
TEST_SIZE = 0.30
RANDOM_STATE = 42

# Load the selected dataset
df = pd.read_csv("breast-cancer-mean-minmax.csv")

X = df.drop(columns=[TARGET])
y = df[TARGET]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

results = []
```

```python
print(
        f"Classifier: KNN with Hyperparameter n_neighbors | "
)
for k in [3, 9, 15, 21]:

    model = KNeighborsClassifier(n_neighbors=k)

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    print(
        f"Value: {k} | "
        f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}% | "
        f"Precision: {precision_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"Recall: {recall_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"F1 Score: {f1_score(y_test, y_pred, pos_label='M'):.4f}"
    )

    results.append({
        "Classifier": "KNN",
        "Hyperparameter": "n_neighbors",
        "Value": k,
        "Accuracy (%)": accuracy_score(y_test, y_pred) * 100,
        "Precision": precision_score(y_test, y_pred, pos_label="M"),
        "Recall": recall_score(y_test, y_pred, pos_label="M"),
        "F1 Score": f1_score(y_test, y_pred, pos_label="M")
    })
```

```python
print(
        f"Classifier: Decision Tree with Hyperparameter max_depth | "
)
for depth in [2, 8, 14]:

    model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=RANDOM_STATE
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    
    print(
        f"Value: {depth} | "
        f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}% | "
        f"Precision: {precision_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"Recall: {recall_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"F1 Score: {f1_score(y_test, y_pred, pos_label='M'):.4f}"
    )

    results.append({
        "Classifier": "Decision Tree",
        "Hyperparameter": "max_depth",
        "Value": depth,
        "Accuracy (%)": accuracy_score(y_test, y_pred) * 100,
        "Precision": precision_score(y_test, y_pred, pos_label="M"),
        "Recall": recall_score(y_test, y_pred, pos_label="M"),
        "F1 Score": f1_score(y_test, y_pred, pos_label="M")
    })
```

```python
print(
        f"Classifier: AdaBoost with Hyperparameter n_estimators | "
)
for n in [10, 20, 30]:

    model = AdaBoostClassifier(
        n_estimators=n,
        random_state=RANDOM_STATE
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    print(
        f"Value: {n} | "
        f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}% | "
        f"Precision: {precision_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"Recall: {recall_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"F1 Score: {f1_score(y_test, y_pred, pos_label='M'):.4f}"
    )

    results.append({
        "Classifier": "AdaBoost",
        "Hyperparameter": "n_estimators",
        "Value": n,
        "Accuracy (%)": accuracy_score(y_test, y_pred) * 100,
        "Precision": precision_score(y_test, y_pred, pos_label="M"),
        "Recall": recall_score(y_test, y_pred, pos_label="M"),
        "F1 Score": f1_score(y_test, y_pred, pos_label="M")
    })
```

```python
print(
        f"Classifier: Random Forest with Hyperparameter n_estimators | "
)
for n in [10, 30, 50, 60]:

    model = RandomForestClassifier(
        n_estimators=n,
        random_state=RANDOM_STATE
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    print(
        f"Value: {n} | "
        f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}% | "
        f"Precision: {precision_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"Recall: {recall_score(y_test, y_pred, pos_label='M'):.4f} | "
        f"F1 Score: {f1_score(y_test, y_pred, pos_label='M'):.4f}"
    )

    results.append({
        "Classifier": "Random Forest",
        "Hyperparameter": "n_estimators",
        "Value": n,
        "Accuracy (%)": accuracy_score(y_test, y_pred) * 100,
        "Precision": precision_score(y_test, y_pred, pos_label="M"),
        "Recall": recall_score(y_test, y_pred, pos_label="M"),
        "F1 Score": f1_score(y_test, y_pred, pos_label="M")
    })
```

```python
results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
```
