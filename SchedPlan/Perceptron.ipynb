```python
import numpy as np
import pandas as pd

# Load data
train = pd.read_csv("data/SeaSynTrain.csv")
test = pd.read_csv("data/SeaSynTest.csv")

X_train = train[["attrib1", "attrib2", "attrib3"]].values
y_train = train["class"].values

X_test = test[["attrib1", "attrib2", "attrib3"]].values
y_test = test["class"].values


class Perceptron:

    def __init__(self, learning_rate=0.001):
        self.learning_rate = learning_rate
        self.weights = None
        self.bias = 0

    def fit(self, X, y, epochs):

        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        for _ in range(epochs):

            for x, target in zip(X, y):

                # Linear output
                z = np.dot(x, self.weights) + self.bias

                # Step activation
                prediction = 1 if z >= 0 else 0

                # Perceptron update
                error = target - prediction

                self.weights += self.learning_rate * error * x
                self.bias += self.learning_rate * error

    def predict(self, X):

        z = np.dot(X, self.weights) + self.bias

        return (z >= 0).astype(int)


def accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)

epochs_list = [1, 5, 10, 15, 20, 50, 60, 80,
               100, 120, 150, 200]
```

```python

results = []

for epochs in epochs_list:

    model = Perceptron(learning_rate=0.01)

    # Train using the training set
    model.fit(X_train, y_train, epochs)

    # Training accuracy
    train_predictions = model.predict(X_train)
    train_acc = accuracy(y_train, train_predictions)

    # Test accuracy
    test_predictions = model.predict(X_test)
    test_acc = accuracy(y_test, test_predictions)

    results.append((epochs, train_acc, test_acc))

    print(
        f"Epochs: {epochs:3d}, "
        f"Training Accuracy: {train_acc:.4f}, "
        f"Test Accuracy: {test_acc:.4f}"
    )
```
